```python
import pandas as pd

# Create a sample dataset
data = {
    'Name': ['Amit', 'Riya', 'Rahul', 'Sneha', 'Amit'],
    'Marks': [85,  None, 72, 90, 85],
    'City': ['Pune', 'Mumbai', None, 'Pune', 'Pune'],
    'Remarks': ['Good', 'Average', 'Good', 'Excellent', 'Good']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Display dataset information
print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Information:")
df.info()

print("\nStatistical Description:")
print(df.describe())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Handle missing values
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['City'] = df['City'].fillna(df['City'].mode()[0])

# Identify duplicate records
print("\nDuplicate Records:")
print(df[df.duplicated()])

# Remove duplicate records
df = df.drop_duplicates()

# Drop irrelevant column
df = df.drop('Remarks', axis=1)

# Select students scoring 75 or above using loc[]
print("\nStudents Scoring 75 or Above:")
print(df.loc[df['Marks'] >= 75])

# Display cleaned dataset
print("\nCleaned Dataset:")
print(df)

# Save cleaned data to CSV file
df.to_csv('cleaned_students.csv', index=False)

print("\nPreprocessed data saved successfully!")
```
